```python
import numpy as np
import pickle

leading_string_1 = "pos_test_refcoeff_1"
leading_string_2 = "pos_test_refcoeff_0.9"
leading_string_3 = "pos_test_refcoeff_0.7"


file = open(f"../rust/out/{leading_string_1}_ffphases.pickle", "rb")
ffphases_1 = pickle.load(file)

file = open(f"../rust/out/{leading_string_2}_ffphases.pickle", "rb")
ffphases_2 = pickle.load(file)

file = open(f"../rust/out/{leading_string_3}_ffphases.pickle", "rb")
ffphases_3 = pickle.load(file)

print(np.shape(ffphases_1))
print(np.shape(ffphases_2))
print(np.shape(ffphases_3))

print(ffphases_1)
print(ffphases_2)
print(ffphases_3)
```

```python
diff12 = np.abs(np.array(ffphases_1) - np.array(ffphases_2))

print(diff12)
print(np.mean(diff12))
```

```python
diff23 = np.abs(np.array(ffphases_2) - np.array(ffphases_3))

print(diff23)
print(np.mean(diff23))
```
